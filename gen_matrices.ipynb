This notebook contains the code by which we created all the test cases for this research work

```python
import numpy as np
import copy
```

### Test case generation Experiment 1

```python
#In general:
#V is n times m
#W is n times k
#H is k times m
#Description: We are generating matrix V for 1 x 1, 2 x 2, 3 x 3, 4 x 4 (with k = 1,2,3,4 respectively)
#Here V is generated by first generating W and H
seedctr = 0
for n in [1,2,3,4]:
    for i in range(100):
        np.random.seed(i)
        W = np.round(np.random.rand(n,n),2) #W has values in [0,1) rounded to two decimal places
        H = np.round(np.random.rand(n,n),2) #H has values in [0,1) rounded to two decimal places
        V = W @ H #Creating V from W and H
        store_dict = {}
        store_dict['V'] = copy.deepcopy(V)
        store_dict['W'] = copy.deepcopy(W)
        store_dict['H'] = copy.deepcopy(H)
        
        filename = "mat_"+str(n)+"x"+str(n)+"_"+str(i)+".npz"
        filepath = "./test_matrices/dirac3_exp1/size_" + str(n) + "x" + str(n) + "/" + filename 
        
        np.savez(filepath,**store_dict)
        print("seed:",seedctr)
        print(V)
        print("---")
        seedctr += 1

```

### Test case generation for Experiment 3

### Test case generation for Experiment 4
Test set B

$4 \times 8$ matrix V factored into $4 \times 3$ W and $3 \times 8$ H
Here we prepare W and H beforehand

```python
acceptable_vals = []
pow = 2
for i in range(2** (pow+1)):
    #acceptable_vals.append(round(1/(2**(pow+1) -1) * i,2)) #simplified the acceptable values list
    acceptable_vals.append(i)
```

```python
acceptable_vals
```

```python
#In general:
#V is n times m
#W is n times k
#H is k times m
#Description: We are generating matrix V for 4 x 8 (with k = 3 respectively)
#Here V is generated by first generating W and H
seedctr = 0
n = 4
k = 3
for m in [8]:
    for i in range(100):
        np.random.seed(seedctr)
        W = np.random.choice(acceptable_vals,(n,k))  #W has int values in 0 to 7 range
        H = np.random.choice(acceptable_vals,(k,m)) #H has int values in 0 to 7 range
        V = W @ H #Creating V from W and H
        store_dict = {}
        store_dict['V'] = copy.deepcopy(V)
        store_dict['W'] = copy.deepcopy(W)
        store_dict['H'] = copy.deepcopy(H)
        
        filename = "mat_"+str(n)+"x"+str(m)+"_"+str(i)+".npz"
        filepath = "./test_matrices/dirac3_exp4a/size_" + str(n) + "x" + str(m) + "/" + filename 
        
        np.savez(filepath,**store_dict)
        print("seed:",seedctr)
        print(V)
        print("---")
        seedctr += 1
```

### Test case generation for Experiment 4
Test set B

```python
n = 4
m = 8
k = 3
max_val = 7
acceptable_vals = [i for i in range(max_val*max_val*k)] #acceptable values go from 0 to the maximum it can be with matrix multiplication for one cell
```

```python
#In general:
#V is n times m
#W is n times k
#H is k times m
#Description: We are generating matrix V for 4 x 8 (with k = 3 respectively)
#Here V is generated without making W and H
seedctr = 0
n = 4
k = 3
for m in [8]:
    for i in range(100):
        np.random.seed(seedctr)
        V = np.random.choice(acceptable_vals,(n,m))  #V has int values in 0 to max_val^2 * k
        filename = "mat_"+str(n)+"x"+str(m)+"_"+str(i)+".npy"
        filepath = "./test_matrices/dirac3_exp4/size_" + str(n) + "x" + str(m) + "/" + filename 
        
        np.save(filepath,V)
        print("seed:",seedctr)
        print(V)
        print("---")
        seedctr += 1
```
